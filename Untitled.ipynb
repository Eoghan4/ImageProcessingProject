```python
import cv2
import numpy as np
import easygui

f = easygui.fileopenbox()
I = cv2.imread(f)

cv2.imshow('I',I)
key = cv2.waitKey(0)

grayscaleClock = cv2.cvtColor(I, cv2.COLOR_BGR2GRAY)
eqgrayscaleClock = cv2.equalizeHist(grayscaleClock)

blurred = cv2.GaussianBlur(eqgrayscaleClock, (5, 5), 0)
cv2.imshow('Grayscale',blurred)
key = cv2.waitKey(0)
```
